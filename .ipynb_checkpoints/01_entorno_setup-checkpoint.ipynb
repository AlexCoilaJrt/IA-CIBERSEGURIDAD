# 01 - Configuración del Entorno de Desarrollo Python

Este cuaderno cubre la configuración del entorno de desarrollo para implementar soluciones de seguridad basadas en IA.

**Contenido:**
- Creación del entorno virtual
- Instalación de bibliotecas principales

## 2.1 Requisitos previos

Se recomienda **Python 3.9 o superior**. Versiones anteriores a 3.6 no son compatibles con varias de las bibliotecas utilizadas en esta guía.

## 2.2 Creación del entorno virtual

Ejecutar en terminal (fuera del notebook):

```bash
# Crear el entorno
python3 -m venv ai_security_env

# Activar en Linux / macOS
source ai_security_env/bin/activate

# Activar en Windows
ai_security_env\Scripts\activate
```

## 2.3 Instalación de bibliotecas

````python
# Instalar todas las dependencias necesarias
# Ejecutar esta celda una sola vez
import subprocess
import sys

paquetes = [
    'numpy',
    'pandas',
    'scikit-learn',
    'tensorflow',
    'keras',
    'matplotlib',
    'seaborn',
    'pefile',
    'requests',
    'joblib',
    'imbalanced-learn',
    'shap',
    'lime'
]

for pkg in paquetes:
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', pkg, '-q'])
    print(f'[OK] {pkg} instalado')
````

### Tabla de bibliotecas principales

| Biblioteca | Función en seguridad |
|---|---|
| NumPy | Operaciones matriciales sobre vectores de características |
| pandas | Carga y preprocesamiento de logs y datasets |
| scikit-learn | Algoritmos clásicos de ML (SVM, árboles, bosques) |
| TensorFlow | Modelos de DL para detección de intrusiones |
| Keras | API de alto nivel sobre TensorFlow |
| pefile | Análisis de ejecutables PE para detección de malware |
| shap | Explicabilidad de modelos (XAI) |
| imbalanced-learn | Manejo de clases desbalanceadas en datasets de seguridad |

## 2.4 Verificación del entorno

````python
import sys
import importlib

REQUIRED = [
    'numpy', 'pandas', 'sklearn', 'tensorflow',
    'keras', 'matplotlib', 'seaborn', 'pefile', 'requests'
]

print(f'Python {sys.version}')
print('=' * 40)
for pkg in REQUIRED:
    try:
        mod = importlib.import_module(pkg)
        version = getattr(mod, '__version__', 'OK')
        print(f'  [OK]    {pkg:<18} {version}')
    except ImportError:
        print(f'  [FALTA] {pkg}')
````
