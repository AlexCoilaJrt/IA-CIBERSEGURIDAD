## 2.3 Instalación de bibliotecas

> **Importante:** Las versiones están fijadas para evitar conflictos entre paquetes.
> - `numpy==1.26.4` es requerido por `tensorflow 2.16.2` (no compatible con numpy >= 2.0)
> - `shap==0.44.1` es la última versión compatible con numpy < 2.0

````python
import subprocess
import sys

# Versiones pinneadas para garantizar compatibilidad
paquetes = [
    'numpy==1.26.4',        # tensorflow 2.16.2 requiere numpy < 2.0
    'pandas',
    'scikit-learn',
    'tensorflow==2.16.2',
    'keras',
    'matplotlib',
    'seaborn',
    'pefile',
    'requests',
    'joblib',
    'imbalanced-learn',
    'shap==0.44.1',         # compatible con numpy < 2.0
    'lime',
    'scipy'
]

print('Instalando dependencias...\n')
errores = []
for pkg in paquetes:
    nombre = pkg.split('==')[0]
    try:
        resultado = subprocess.run(
            [sys.executable, '-m', 'pip', 'install', pkg, '-q'],
            capture_output=True, text=True
        )
        if resultado.returncode == 0:
            print(f'  [OK] {nombre}')
        else:
            print(f'  [ERROR] {nombre}')
            errores.append(nombre)
    except Exception as e:
        print(f'  [FALLO] {nombre}: {e}')
        errores.append(nombre)

print()
if errores:
    print(f'Paquetes con error: {errores}')
else:
    print('Todas las dependencias instaladas correctamente.')
````

## 2.4 Verificación del entorno

````python
import sys
import importlib

REQUIRED = [
    ('numpy',      '1.26.4 requerido por tensorflow'),
    ('pandas',     ''),
    ('sklearn',    ''),
    ('tensorflow', '2.16.2 requerido'),
    ('keras',      ''),
    ('matplotlib', ''),
    ('seaborn',    ''),
    ('pefile',     ''),
    ('requests',   ''),
    ('joblib',     ''),
    ('imblearn',   ''),
    ('shap',       '0.44.1 requerido'),
    ('scipy',      ''),
]

print(f'Python {sys.version}')
print('=' * 55)
todo_ok = True
for pkg, nota in REQUIRED:
    try:
        mod = importlib.import_module(pkg)
        version = getattr(mod, '__version__', 'OK')
        info = f'  ({nota})' if nota else ''
        print(f'  [OK]    {pkg:<18} {version}{info}')
    except ImportError:
        print(f'  [FALTA] {pkg}')
        todo_ok = False

print('=' * 55)
if todo_ok:
    print('Entorno listo. Puedes ejecutar los notebooks 02 al 08.')
else:
    print('Algunos paquetes faltan. Vuelve a ejecutar la celda de instalación.')
````
